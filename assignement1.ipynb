```python
employee = {101: {'name': 'Satya', 'age': 27, 'department': 'HR', 'salary': 50000}}
```

```python
employee
```

```python
def add_employee():
    emp_id = int(input("enter the employee id"))
    name = str(input("enter the employee name"))
    age = int(input("enter the employee age"))
    department = str(input("enter the employee department"))
    salary = int(input("enter the employee salary"))

    employee[emp_id] = {
          'name' : name, 'age' : age, 'department' : department,'salary' : salary
    }
```

```python
employee
```

```python
def display_employee():
    for emp_id,v in employee.items():
        print(emp_id)
        for i,j in v.items():
            print( i,"---->",j)
```

```python
def search_employee():
    emp_id = int(input("enter the employee id"))
    if emp_id in employee:
            print("name","--->",employee[emp_id]['name'])
            print("age",':',employee[emp_id]['age'])
            print("department" ,':',employee[emp_id]['department'])
            print("salary",':',employee[emp_id]['salary'])
    else:
        print("not found data")
```

```python
def exit():
    print(" thank you message")
    return
```

```python
def main_menu():
    while  True:
        choice = int(input("enter the user choice"))
        print("if choice is 1 : add the employee in the system")
        print("if choice is 2 : display all the employee in the system")
        print("if choice is 3 : search the employee in the system")
        print("if choice is 4 : exit from the program")
        if choice == 1:
            add_employee()
        elif choice == 2:
            display_employee()
        elif choice ==3:
            search_employee()
        elif choice == 4:
            exit()
            break
        else:
            print("please enter the valid choice")
        
        
    
```

```python
main_menu()
```
